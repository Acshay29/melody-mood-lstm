## 1 · Train your next-note predictor
Run the same environment in which `torch`, `numpy`, and `Pillow` are installed. CPU works for a small trial; use more epochs for genuine training.

```
from pathlib import Path
import sys, subprocess
root = Path.cwd()
project = root / "src" / "project.py"
assert project.exists(), "Open this notebook from the kit root directory"
subprocess.run([sys.executable, str(project), "--mode", "train", "--epochs", "12", "--examples", "400", "--output-dir", "outputs/training"], check=True)
```
